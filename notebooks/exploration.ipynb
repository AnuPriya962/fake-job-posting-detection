```python
import sys
import os
from pathlib import Path
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Add root directory to sys.path to allow imports from src/
sys.path.append(str(Path.cwd().parent))

# Import your data loader functions
from src.data_loader import load_dataset, validate_schema, clean_dataset, TARGET_COLUMN
```

```python
# Load dataset relative to root folder
dataset_path = "../data/fake_job_postings.csv"

raw_df = load_dataset(dataset_path)
validate_schema(raw_df)
df = clean_dataset(raw_df)

print(f"\nFinal Cleaned Dataset Shape: {df.shape}")
```

```python
plt.figure(figsize=(7, 4))
ax = sns.countplot(data=df, x=TARGET_COLUMN, palette=["#2b5c8f", "#d9534f"])
plt.xticks(ticks=[0, 1], labels=["Real / Legitimate", "Fake / Fraudulent"])
plt.title("Class Distribution: Real vs Fraudulent Job Postings")
plt.xlabel("Posting Authenticity")
plt.ylabel("Count")

total = len(df)
for p in ax.patches:
    percentage = f'{100 * p.get_height() / total:.2f}%'
    ax.annotate(f'{p.get_height()}\n({percentage})', 
                (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', color='white', fontweight='bold')
plt.show()
```

```python
df['description_len'] = df['description'].apply(len)

plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x=TARGET_COLUMN, y='description_len', palette=["#2b5c8f", "#d9534f"])
plt.xticks(ticks=[0, 1], labels=["Real / Legitimate", "Fake / Fraudulent"])
plt.title("Job Description Character Length Distribution")
plt.xlabel("Posting Authenticity")
plt.ylabel("Character Length (Log Scale)")
plt.yscale('log')
plt.show()
```
